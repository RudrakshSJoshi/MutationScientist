```python
import os
import socket
import subprocess

print("Hostname:", socket.gethostname())
print("FQDN:", socket.getfqdn())
print("/usr/bin/nvidia-smi exists:", os.path.exists("/usr/bin/nvidia-smi"))
print("/dev/nvidia0 exists:", os.path.exists("/dev/nvidia0"))

result = subprocess.run(
    ["/usr/bin/nvidia-smi"],
    capture_output=True,
    text=True
)

print("Return code:", result.returncode)
print(result.stdout)
print(result.stderr)
```
